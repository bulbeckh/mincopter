### MinCopter Thrust Testing

```python
from matplotlib import pyplot as plt
import numpy as np

## measurements
x = [1100, 1200, 1300, 1400, 1500, 1600, 1700, 1800]
mass_readings = [2240, 2185, 2030, 1855, 1730, 1600, 1510, 1300]
y = [2240 - e for e in mass_readings]

m,b = np.polyfit(x,y,1)
print(m,b)

plt.plot(np.linspace(1100,1800,100), m*np.linspace(1100,1800,100) + b, color='purple', linestyle='--')
plt.plot(x,y, label='thrust line')
plt.axhline(y=0, color='black', linestyle=':')
plt.axhline(y=820, color='red', linestyle='--', label='Frame+Battery weight')
plt.legend()
plt.xlabel('PWM')
plt.ylabel('mass (g)')
plt.show()
```
